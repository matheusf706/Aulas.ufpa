Questão 4.1: Uma memória RAM (MP) é fabricada com a possibilidade de armazenar um máximo de 256k bits. Cada célula pode armazenar 8 bits. Qual é o tamanho de cada endereço e qual é o total de células que podem ser utilizadas naquela RAM?

```python
import math

T = 2**1 * 2**10 * 2**4
KB = T // (1024 * 8)
expoente = math.log2(T)

print(f"{KB}KB")

print(f"É igual a 2^{int(expoente)}")
```

Exemplo 4.2:
Uma memória RAM (MP) é fabricada com a possibilidade de armazenar um máximo de 256K bits. Cada célula pode armazenar 8 bits. Qual é o tamanho de cada endereço e qual é o total de células que podem ser utilizadas naquela RAM?

```python
import math

def calcular_ram(capacidade_bits, bits_por_celula):
    total_celulas = capacidade_bits / bits_por_celula

    tamanho_endereco = math.ceil(math.log2(total_celulas))

    return int(total_celulas), tamanho_endereco

capacidade_k_bits = 256 * 1024
celula_tamanho = 8

total_celulas, tamanho_endereco=calcular_ram(capacidade_k_bits, celula_tamanho)

print(f"Total de células utilizáveis: {total_celulas}")
print(f"Tamanho de cada endereço: {tamanho_endereco} bits")
```

Exemplo 4.3:
Um computador, cuja memória RAM (MP) tem uma capacidade máxima de armazenamento de 2K palavras de 16 bits cada, possui um REM e um RDM. Qual o tamanho desses registradores; qual o valor do maior endereço dessa MP e qual a quantidade total de bits que nela podem ser armazenados?

```python
import math

def resolução(capacidade_k_palavras, bits_por_palavra):
    total_palavras = capacidade_k_palavras * 1024

    tamanho_rem = math.ceil(math.log2(total_palavras))

    tamanho_rdm = bits_por_palavra

    maior_endereco = total_palavras - 1

    total_bits = total_palavras * bits_por_palavra

    return tamanho_rem, tamanho_rdm, maior_endereco, total_bits

k_palavras = 2
bits_palavra = 16

rem, rdm, maior_end, total_bits = resolução(k_palavras, bits_palavra)

print(f"Tamanho do REM: {rem} bits")
print(f"Tamanho do RDM: {rdm} bits")
print(f"Maior endereço da MP (em decimal): {maior_end}")
print(f"Maior endereço da MP (em binário): {bin(maior_end)[2:]}")
print(f"Quantidade total de bits: {total_bits} bits")
```

Exemplo 4.4:
Um processador possui um RDM com capacidade de armazenar 32 bits e um REM com capacidade de armazenar 24 bits. Sabendo-se que em cada acesso são lidas duas células da memória RAM (MP) e que o barramento de dados (BD) tem tamanho igual ao da palavra, pergunta-se:                                                      
a) Qual é a capacidade máxima de endereçamento do microcomputador em questão?    
b) Qual é o total máximo de bits que podem ser armazenados na memória RAM (MP)?  
c) Qual é o tamanho da palavra e de cada célula da máquina?

```python
import math

bits_rdm = 32
bits_rem = 24
celulas_por_acesso = 2

print("=== RESOLUÇÃO DO EXEMPLO 4.4 ===\n")

# --- Letra A ---
capacidade_endereçamento = 2 ** bits_rem
capacidade_endereçamento_k = capacidade_endereçamento / 1024
capacidade_endereçamento_m = capacidade_endereçamento_k / 1024

print(f"a) Capacidade máxima de endereçamento:")
print(f"   - {capacidade_endereçamento} posições de memória.")
print(f"   - Ou seja, {int(capacidade_endereçamento_m)}M de posições de memória.\n")


# --- Letra C (calculada antes da B por dependência de dados) ---
tamanho_palavra = bits_rdm

tamanho_celula = tamanho_palavra / celulas_por_acesso

print(f"c) Tamanhos da palavra e da célula:")
print(f"   - Tamanho da palavra: {tamanho_palavra} bits.")
print(f"   - Tamanho de cada célula: {int(tamanho_celula)} bits.\n")


# --- Letra B ---

total_bits_ram = capacidade_endereçamento * tamanho_celula
total_bytes_ram = total_bits_ram / 8
total_megabytes_ram = total_bytes_ram / (1024 * 1024)

print(f"b) Total máximo de bits armazenados na RAM:")
print(f"   - {int(total_bits_ram)} bits.")
print(f"   - Convertendo para uma unidade comum: {int(total_megabytes_ram)} MB (Megabytes).")
```

Exemplo 4.5:
Um processador possui um BE (barramento de endereços) com capacidade de permitir a transferência de 33 bits de cada vez. Sabe-se que o BD (barramento de dados) permite a transferência de quatro palavrasem cada acesso e que cada célula da memória RAM (MP) armazena um oitavo (1/8) de cada palavra. Considerando que a memória RAM (MP) pode armazenar um máximo de 64G bits, pergunta-se:                                                                     
a) Qual é a quantidade máxima de células que podem ser armazenadas na memória RAM (MP)?                                                                        
b) Qual é o tamanho do REM e do BD existentes neste processador?                 
c) Qual é o tamanho de cada célula e da palavra desta máquina?

```python
import math


bits_be = 33
palavras_por_acesso = 4
proporcao_celula_palavra = 1 / 8
max_capacidade_bits = 64 * (1024 * 1024 * 1024)

print("=== RESOLUÇÃO CORRETA DO EXEMPLO 4.5 ===\n")

# --- Letra A ---
quantidade_max_celulas = 2 ** bits_be
quantidade_celulas_g = quantidade_max_celulas / (1024 * 1024 * 1024)

print(f"a) Quantidade máxima de células na memória RAM (MP):")
print(f"   - {quantidade_max_celulas} células.")
print(f"   - Ou seja, {int(quantidade_celulas_g)}G células.\n")


# --- Letra C ---
tamanho_celula_bits = max_capacidade_bits / quantidade_max_celulas

tamanho_palavra_bits = tamanho_celula_bits / proporcao_celula_palavra

print(f"c) Tamanho de cada célula e da palavra:")
print(f"   - Tamanho de cada célula: {int(tamanho_celula_bits)} bits.")
print(f"   - Tamanho da palavra: {int(tamanho_palavra_bits)} bits.\n")


# --- Letra B ---
tamanho_rem = bits_be

tamanho_bd_bits = palavras_por_acesso * tamanho_palavra_bits

print(f"b) Tamanho do REM e do BD:")
print(f"   - Tamanho do REM: {tamanho_rem} bits.")
print(f"   - Tamanho do BD: {int(tamanho_bd_bits)} bits.")
```

1) Um computador possui uma memória principal com capacidade para armazenar palavras de 16 bits em cada uma de suas N células, e o seu barramento de endereços tem 12 bits de tamanho. Sabendo-se que em cada célula pode-se armazenar o valor exato de uma palavra, quantos bytes poderão ser armazena-dos nessa memória?

```python
bits_por_palavra = 16
tamanho_barramento = 12

bytes_por_palavra = bits_por_palavra / 8

N = 2**tamanho_barramento

capacidade_total_bytes = N * bytes_por_palavra

print(f"Total de células (N): {N}")
print(f"Tamanho da palavra em bytes: {bytes_por_palavra}")
print(f"Capacidade total da memória: {capacidade_total_bytes}kbytes")
```
